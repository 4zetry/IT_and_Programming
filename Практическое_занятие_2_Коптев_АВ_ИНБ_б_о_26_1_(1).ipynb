## Уровень 1 – Базовый, вариант 8

Пользователь вводит целое число n. Вычислить факториал с помощью цикла `for` (без `math.factorial`).

```python
n = int(input("Введите целое n: "))

if n < 0:
    print("Факториал определён только для неотрицательных чисел")
else:
    result = 1
    for i in range(2, n + 1):
        result *= i
    print(f"{n}! = {result}")
```

```python
# дополнительная проверка граничных случаев: 0! и 10!
for n in (0, 1, 10):
    result = 1
    for i in range(2, n + 1):
        result *= i
    print(f"{n}! = {result}")
```

## Уровень 2 – Средний, вариант 8

С помощью цикла `while` и операции `%` определить, является ли запись числа n в двоичной системе палиндромом. Вывести «да» или «нет».

```python
def binary_digits(n):
    """Двоичные цифры числа n (старшая — первая), получаем через % и //."""
    if n == 0:
        return [0]
    bits = []
    while n > 0:
        bits.append(n % 2)   # очередная младшая цифра
        n //= 2
    bits.reverse()
    return bits

def is_binary_palindrome(n):
    bits = binary_digits(n)
    left, right = 0, len(bits) - 1
    while left < right:                  # сравниваем цифры с краёв к центру
        if bits[left] != bits[right]:
            return False
        left += 1
        right -= 1
    return True

n = int(input("Введите неотрицательное целое n: "))
if n < 0:
    print("Число должно быть неотрицательным")
else:
    print("Двоичная запись:", "".join(map(str, binary_digits(n))))
    print("да" if is_binary_palindrome(n) else "нет")
```

```python
# проверка на нескольких значениях
for n in (0, 1, 5, 6, 9, 10, 27, 33):
    print(n, "".join(map(str, binary_digits(n))), "да" if is_binary_palindrome(n) else "нет")
```

## Уровень 3 – Продвинутый, вариант 8

Вычислить ln(2) двумя способами и сравнить скорость сходимости для одного и того же числа членов n:

1. ряд ln 2 = 1 − 1/2 + 1/3 − 1/4 + …
2. ряд ln((1+x)/(1−x)) = 2(x + x³/3 + x⁵/5 + …) при x = 1/3. Так как (1 + 1/3)/(1 − 1/3) = 2, получаем ln 2 = 2·Σ x^(2k+1)/(2k+1).

```python
import math

def ln2_alternating(n):
    """Первые n членов ряда 1 - 1/2 + 1/3 - ..."""
    total = 0.0
    sign = 1
    for k in range(1, n + 1):
        total += sign / k
        sign = -sign
    return total

def ln2_atanh(n, x=1/3):
    """Первые n членов ряда 2*(x + x^3/3 + x^5/5 + ...)"""
    total = 0.0
    for k in range(n):
        total += x ** (2 * k + 1) / (2 * k + 1)
    return 2 * total

n = int(input("Введите число членов ряда n: "))
r1 = ln2_alternating(n)
r2 = ln2_atanh(n)
exact = math.log(2)

print(f"n = {n}")
print(f"Ряд 1:      {r1:.12f}   ошибка {abs(r1 - exact):.3e}")
print(f"Ряд 2:      {r2:.12f}   ошибка {abs(r2 - exact):.3e}")
print(f"math.log(2) {exact:.12f}")
```

```python
# сравнение скорости сходимости для разных n
print(f"{'n':>4} | {'ошибка ряда 1':>14} | {'ошибка ряда 2':>14}")
print("-" * 38)
for n in (1, 2, 5, 10, 20, 50, 100):
    e1 = abs(ln2_alternating(n) - math.log(2))
    e2 = abs(ln2_atanh(n) - math.log(2))
    print(f"{n:>4} | {e1:>14.3e} | {e2:>14.3e}")
```

**Вывод.** Ряд 1 (знакочередующийся) сходится очень медленно: погрешность убывает примерно как 1/(2n), т. е. для каждого нового верного знака нужно примерно в 10 раз больше членов (при n = 100 ошибка всё ещё около 5·10⁻³). Ряд 2 при x = 1/3 сходится геометрически: каждый следующий член уменьшается примерно в 9 раз (x² = 1/9), поэтому уже при n = 10 ошибка около 10⁻¹¹, а при n = 20 результат совпадает с `math.log(2)` до предела точности `float` (≈ 2·10⁻¹⁶).

### 1. Отличие `for` и `while`

`for` — цикл по известной последовательности/заданному числу повторений (перебор элементов, `range`). `while` — цикл «пока верно условие», когда число итераций заранее неизвестно.

```python
# for: сумма чисел от 1 до 10 — число итераций известно
total = 0
for i in range(1, 11):
    total += i
print(total)

# while: ввод пароля до правильного — число попыток неизвестно
attempts = ["1", "abc", "secret"]
i = 0
while attempts[i] != "secret":
    i += 1
print("Пароль принят с попытки", i + 1)
```

### 2. Блок `else` в цикле

`else` выполняется, если цикл завершился **естественно** (исчерпан итератор или условие `while` стало ложным), и **не** выполняется, если цикл прерван `break`.

```python
for i in range(2, 5):
    if i == 10:
        break
else:
    print("else сработал: break не было")

for i in range(2, 5):
    if i == 3:
        break
else:
    print("этот else НЕ выведется — был break")
print("конец")
```

### 3. `break` и `continue`

```python
i = 0
while i < 10:
    i += 1
    if i % 2 == 0:
        continue          # чётные: пропускаем остаток тела, переходим к следующей итерации
    if i > 7:
        break             # при i = 9 выходим из цикла полностью
    print("итерация, i =", i)   # напечатает 1, 3, 5, 7
print("после цикла i =", i)
```

### 4. Типичные ошибки в циклах

```python
# 1) Бесконечный цикл: забыли изменить счётчик
#   i = 0
#   while i < 3:
#       print(i)          # i не меняется!
i = 0
while i < 3:              # исправлено
    print(i, end=" ")
    i += 1
print()

# 2) Неверный диапазон range: верхняя граница не включается
print(list(range(1, 5)))      # ошибка: ожидали 1..5
print(list(range(1, 5 + 1)))  # исправлено

# 3) Неверное условие выхода (off-by-one): while i <= len(a) даст IndexError
a = [10, 20, 30]
i = 0
while i < len(a):             # правильно: строго <
    print(a[i], end=" ")
    i += 1
print()
```

### 5. Конструкция `if / elif / else`

```python
def classify(x):
    if x < 0:
        return "меньше 0"
    elif x <= 10:
        return "от 0 до 10"
    else:
        return "больше 10"

for v in (-5, 0, 7, 10, 11):
    print(v, "->", classify(v))
```

Порядок условий важен: проверка идёт сверху вниз и выполняется только первая истинная ветка. Если поставить `x <= 10` первым, отрицательные числа попали бы в ветку «от 0 до 10».

### 6. Логические операторы `and`, `or`, `not`

`and` — истина, если истинны оба операнда; `or` — если хотя бы один; `not` — инверсия. Приоритет: `not` > `and` > `or`.

```python
def check(n):
    return (n % 2 == 0 and 10 <= n <= 20) or (n % 2 == 1 and n > 100)

for v in (12, 22, 101, 99, 100, 20):
    print(v, check(v))
```

### 12. Три способа вычисления квадратного корня

```python
x = 2
print(math.sqrt(x))         # 1.4142135623730951
print(x ** 0.5)             # 1.4142135623730951

y = x
while True:                 # метод Герона
    y_new = (y + x / y) / 2
    if abs(y_new - y) < 1e-12:
        break
    y = y_new
print(y_new)
```

| Способ | Плюсы | Минусы |
|---|---|---|
| `math.sqrt()` | быстрый, точный, понятный; ошибка на отрицательных | нужен `import`; `ValueError` для x < 0 |
| `x ** 0.5` | без импорта, работает в выражениях | для отрицательных x возвращает комплексное число; на больших целых возможна потеря точности |
| Метод Герона | показывает алгоритм, управляемая точность, работает на любых числах, где задан делитель | медленнее, нужен критерий остановки и контроль x > 0 |

### 14. Почему `input()` нужно преобразовывать

```python
s = input("Введите число: ")      # всегда str
try:
    print(s + 1)                   # str + int -> TypeError
except TypeError as e:
    print("TypeError:", e)

try:
    number = int(s)               # правильно
    print(number + 1)
except ValueError:
    print("Введено не целое число")
```

```python
s = input("Введите число: ")
try:
    number = int(s)
    print(number + 1)
except ValueError:
    print("Введено не целое число")
```

`input()` всегда возвращает `str`, поэтому для арифметики нужны `int()` или `float()`, а для пользовательского ввода — защита `try/except ValueError`.

### 15. Цикл и встроенные функции

```python
data = [7, 3, 9, 1, 5]

# вариант 1: цикл с аккумуляцией
total = 0
minimum = data[0]
for v in data:
    total += v
    if v < minimum:
        minimum = v
print(total, minimum)

# вариант 2: встроенные функции
print(sum(data), min(data))
```

Версия со встроенными функциями короче (1 строка вместо 6), читаемее и надёжнее (меньше места для ошибок, реализация на C — быстрее). Цикл нужен, когда требуется нестандартная логика (несколько величин за один проход, условия, ранний выход).
