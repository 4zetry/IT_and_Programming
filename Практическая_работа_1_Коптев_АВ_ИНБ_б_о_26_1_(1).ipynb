Необходимо вычислить значение выражения:
$$AO=\frac{35}{G_1} * Zvcw + G_1 * a - \frac{a + Zvcw}{a}$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $G_1, Zvcw, a. Переменная, которая должна быть вычислена на основе независимых: $$AO$.

```python
G_1=10
Zvcw=30.67
a=5
print("G_1 = ", G_1, ", type(G_1) = ", type(G_1))
print("Zvcw = ", Zvcw, ", type(Zvcw) = ", type(Zvcw))
print("a = ", a, ", type(a) = ", type(a))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $AO$:

```python
AO=(G_1**Zvcw-G_1**(1/3))/(a*(a+120*abs(G_1-Zvcw)))
print('AO=',AO)
print('Type of AO: ', type(AO))
```
