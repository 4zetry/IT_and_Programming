## Задание

Ваня составляет 6-буквенные слова из букв слова КОМПЬЮТЕР так, что в них
можно убрать три буквы и получить слово КОТ. Сколько различных слов может
составить Ваня?

## Решение

```python
import itertools

# Алфавит
alphabet = 'КОМПЬЮТЕР'
target = 'КОТ'

count = 0

# Генерируем все возможные 6-буквенные слова
# product перебирает все варианты с повторениями (9^6 = 531441 вариантов, что быстро)
for word_tuple in itertools.product(alphabet, repeat=6):
    word = ''.join(word_tuple)

    # Проверяем, является ли 'КОТ' подпоследовательностью слова
    target_idx = 0
    for char in word:
        if char == target[target_idx]:
            target_idx += 1
            # Если прошли по всему слову и нашли все буквы 'К', 'О', 'Т' по порядку
            if target_idx == len(target):
                break

    # Если нашли полное совпадение, увеличиваем счетчик
    if target_idx == len(target):
        count += 1

print(f"Количество подходящих слов: {count}")
```
