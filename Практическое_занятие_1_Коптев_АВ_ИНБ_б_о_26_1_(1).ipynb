## Уровень 2 – Средний, задание 8

Даны `a = 5`, `b = 2.0`. Вычислить `a + b`, `a * b`, `a / b`, вывести результаты и типы, объяснить, почему тип меняется.

```python
a = 5
b = 2.0

summa = a + b
product = a * b
quotient = a / b

print(f"a + b = {summa}, тип: {type(summa)}")
print(f"a * b = {product}, тип: {type(product)}")
print(f"a / b = {quotient}, тип: {type(quotient)}")
```

**Пояснение.** `a` — `int`, `b` — `float`. Если в операции участвуют `int` и `float`, Python выполняет *неявное приведение*: `int` преобразуется к `float`, потому что `float` — «более широкий» тип (он способен представить дробную часть, а `int` её терять не должен). Поэтому `a + b` и `a * b` дают `float` (7.0 и 10.0), хотя числовые значения целые. Операция `/` (истинное деление) в Python 3 возвращает `float` всегда, даже для двух `int` (например, `6 / 3 = 2.0`).

## Уровень 3 – Продвинутый, задание 8

Дано `number`. Вычислить сумму его цифр (через строковое преобразование и перебор символов, каждый символ приводится обратно в `int`). Вывести результат и тип.

```python
number = int(input("Введите целое число: "))

digits_sum = 0
for ch in str(abs(number)):      # abs() — чтобы знак «-» не попал в перебор
    digits_sum += int(ch)        # символ-цифра -> int

print(f"Сумма цифр числа {number}: {digits_sum}")
print(f"Тип результата: {type(digits_sum)}")
```

Проверка: 4 + 8 + 2 + 7 + 1 = 22. Тип результата — `int`, так как складываются целые числа.

## Контрольные вопросы

### 1. Что такое переменная в Python? Присваивание, примеры имён

Переменная — это имя, связанное со ссылкой на объект в памяти. Оператор `=` не «кладёт значение в ячейку», а привязывает имя к объекту, который создаётся в правой части. Одному объекту могут быть привязаны несколько имён, а имя можно перепривязать к другому объекту.

Правила имён: буквы, цифры и `_`; нельзя начинать с цифры; нельзя использовать ключевые слова (`if`, `for`, `class` …); регистр важен (`age` и `Age` — разные имена).

```python
x = 10            # корректно
user_name = "Анна" # корректно (snake_case)
_count2 = 3       # корректно

# 2count = 5      # некорректно: имя начинается с цифры -> SyntaxError
# for = 1         # некорректно: for — ключевое слово
# my-var = 1      # некорректно: дефис воспринимается как минус
print(x, user_name, _count2)
```

### 2. Основные простейшие типы данных, `type()`, динамическая типизация

```python
for literal in (42, 3.14, "текст", True, None, 2 + 3j):
    print(f"{literal!r:10} -> {type(literal)}")
```

Основные типы: `int`, `float`, `str`, `bool` (а также `NoneType`, `complex`). Типизация **динамическая**: тип принадлежит не имени, а объекту-значению; проверяется он во время выполнения, поэтому одно и то же имя можно сначала привязать к числу, затем к строке.

### 3. Как работает `print()`

```python
name, age = "Анна", 19
print(name)                                   # одна переменная
print("Имя:", name, "Возраст:", age)          # несколько значений через запятую (разделитель — пробел)
print(f"Имя: {name}, возраст: {age}")         # f-строка
print("Имя: " + name + ", возраст: " + str(age))  # конкатенация (число нужно привести к str)
```

Запятая сама приводит значения к строкам и вставляет пробелы (`sep=" "`), конкатенация `+` работает только со строками (иначе `TypeError`), f-строка наиболее удобна при смешении текста и значений и позволяет форматирование (`{x:.2f}`).

### 4. Функция `type()`

```python
number, text, flag = 7, "Привет", False
print(type(number), type(text), type(flag))
```

`type(x)` возвращает класс объекта. Запись `<class 'int'>` означает, что объект принадлежит классу `int`. Типы `type(5)` и `type(5.0)` различны (`int` и `float`), потому что тип определяется формой литерала: наличие десятичной точки создаёт `float`.

### 5. Арифметические операции и типы результатов

```python
print(7 + 2, type(7 + 2))        # int + int -> int
print(7 - 2.0, type(7 - 2.0))    # int - float -> float
print(7 * 2, type(7 * 2))        # int
print(7 / 2, type(7 / 2))        # / всегда float
print(6 / 3, type(6 / 3))        # 2.0, всё равно float
print(7 // 2, type(7 // 2))      # int // int -> int
print(7.0 // 2, type(7.0 // 2))  # float // int -> float
print(7 % 2, type(7 % 2))        # остаток, int
print(2 ** 10, type(2 ** 10))    # int ** int (неотриц. степень) -> int
print(2 ** -1, type(2 ** -1))    # отриц. степень -> float
```

### 6. Деление `/` и `//`

```python
print(7 / 2, 9 / 3)         # положительные: 3.5  3.0
print(7 // 2, 9 // 3)       # положительные: 3  3
print(-7 / 2, -9 / 3)       # отрицательные: -3.5 -3.0
print(-7 // 2, 7 // -2)     # отрицательные: -4 -4 (округление вниз, к -inf)
```

`/` — истинное деление, результат `float`. `//` — целочисленное (floor) деление: результат округляется **вниз** к ближайшему меньшему целому. Поэтому `-7 // 2 = -4`, а не `-3` (это не отбрасывание дробной части).

### 7. Преобразование типов

```python
print(int("123") + 1, float("3.5") * 2, str(42) + "!")
print(int(3.99), int(True), float(7), bool(0), bool("a"))

for s in ("123", "abc", "3.5"):
    try:
        print(f"int({s!r}) =", int(s))
    except ValueError as e:
        print(f"int({s!r}) -> ошибка ValueError: {e}")
```

Ошибка `ValueError` возникает, когда строка не является корректной записью числа (`"abc"`, а также `"3.5"` для `int`). Для безопасного преобразования используют `try/except`.

### 8. Выражение, возвращающее значение

```python
arith = 3 + 4 * 2            # арифметическое
cmp = 10 > 5                 # сравнение
logic = (3 > 1) and not (2 > 5)  # логическое
print(arith, cmp, logic)
print(5 * 5, 4 == 4.0, True or False)   # результат можно сразу передать в print
```

Выражение при вычислении всегда даёт значение, которое можно присвоить переменной, передать в функцию или использовать в другом выражении.

### 9. Операторы сравнения и логические операторы

```python
x = 15
print(x > 10 and x < 20)         # True
print(x < 10 or x % 5 == 0)      # True
print(not x == 15 or x != 3)     # (not (x == 15)) or (x != 3) -> False or True -> True
print(1 < x <= 15)               # цепочка сравнений -> True
```

Приоритет (от высшего к низшему): арифметика → сравнения → `not` → `and` → `or`. Операторы `and`/`or` вычисляются «лениво» (останавливаются, как только результат ясен).

### 10. Операнды разных типов

```python
print(5 + 2.5, type(5 + 2.5))      # неявно: int -> float
print(True + 1, type(True + 1))    # bool -> int
print("ab" * 3)                    # str * int допустимо — повторение
try:
    print("5" + 3)                 # str + int — автоматического приведения нет
except TypeError as e:
    print("TypeError:", e)
print("5" + str(3), int("5") + 3)  # явное приведение
```

### 11. Несколько значений в `print()`: способы

```python
a, b = 3, 4
print("Сумма", a, "+", b, "=", a + b)                 # через запятую — быстро, без форматирования
print(f"Сумма {a} + {b} = {a + b}")                   # f-строка — самый читаемый
print("Сумма {} + {} = {}".format(a, b, a + b))       # .format() — шаблон отдельно от данных
print("Сумма " + str(a) + " + " + str(b) + " = " + str(a + b))  # конкатенация — громоздко
```

### 12. Порядок выполнения операций

```python
print(2 + 3 * 4 ** 2)        # 4**2=16 -> 3*16=48 -> 2+48 = 50
print((2 + 3) * 4 ** 2)      # скобки: 5 * 16 = 80
print(((2 + 3) * 4) ** 2)    # 20 ** 2 = 400
print(2 ** 3 ** 2)           # ** справа налево: 3**2=9 -> 2**9 = 512
```

Приоритет: `**` → унарный минус → `* / // %` → `+ -`. Изменить порядок можно скобками.

### 13. Множественное присваивание

```python
a, b, c = 1, 2, 3          # одновременное присваивание
x = y = z = 0              # одно значение нескольким именам
a, b = b, a                # обмен без временной переменной
first, *rest = [10, 20, 30, 40]   # распаковка списка
p, q = (5, 6)                     # распаковка кортежа
print(a, b, c, x, y, z, first, rest, p, q)
```

### 14. Изменяемые и неизменяемые типы

```python
s = "кот"
print(id(s))
s = s + "а"          # создаётся НОВАЯ строка
print(id(s), s)

n1 = 5
print(id(n1))
n1 += 1              # новое число-объект
print(id(n1))

try:
    s[0] = "К"       # изменить символ на месте нельзя
except TypeError as e:
    print("TypeError:", e)
```

Числа, строки, `bool`, кортежи неизменяемы: «изменение» создаёт новый объект, о чём говорит изменение `id()`. Списки, словари, множества — изменяемые.

### 15. Мини-программа с пояснением строк

```python
price = 49.99                          # 1: float
quantity = 3                           # 2: int
total = price * quantity               # 3: int * float -> float (149.97)
print(f"Итого: {total:.2f} руб.")      # 4: форматирование до 2 знаков
print(type(price), type(quantity), type(total))  # 5: float, int, float
count_str = str(quantity)              # 6: int -> str
print(count_str + " шт.", type(count_str))       # 7: конкатенация строк, тип str
```

Строки 1–2 создают переменные типов `float` и `int`; строка 3 вычисляет произведение (при смешении `int` и `float` результат — `float`); строка 4 выводит результат с двумя знаками после запятой; строка 5 проверяет типы; строки 6–7 демонстрируют явное преобразование в `str` и конкатенацию.
