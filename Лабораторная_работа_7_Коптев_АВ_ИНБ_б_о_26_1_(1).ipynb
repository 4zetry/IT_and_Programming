# Подготовка данных

````python
F = open('p7_data_08.csv', encoding='utf-8')

header = F.readline().strip().split(',')

all_domains = []

for line in F:
    data = line.strip().split(',')

    domain_data = {}

    for i, key in enumerate(header):
        domain_data[key] = data[i]

    price_str = domain_data['price'].replace('$', '').strip()
    domain_data['price'] = float(price_str) if price_str else 0.0

    all_domains.append(domain_data)

F.close()

print(f"Загружено записей: {len(all_domains)}")
print("Пример записи (первая строка):")
print(all_domains[0])
````

## Задание 1 (количество сотрудников без логина)

````python
count_missing_both = 0
count_missing_domain = 0
count_missing_user = 0
count_either = 0

for item in all_domains:
    has_user = bool(item['user'].strip())
    has_domain = bool(item['domain'].strip())

    if not has_user and not has_domain:
        count_missing_both += 1
        count_either += 1
    elif not has_domain:
        count_missing_domain += 1
        count_either += 1
    elif not has_user:
        count_missing_user += 1
        count_either += 1

print(f"Доменов без имени (domain): {count_missing_domain}")
print(f"Доменов без пользователя (user): {count_missing_user}")
print(f"Доменов без обоих полей: {count_missing_both}")
print(f"Всего доменов, где не указан domain или user (или оба): {count_either}")
````

## Задание 2 (средняя цена домена по каждой из 4-х категорий)

````python
categories_stats = {}

for item in all_domains:
    cat = item['Category'].strip()
    if cat:
        if cat not in categories_stats:
            categories_stats[cat] = {'sum': 0.0, 'count': 0}

        categories_stats[cat]['sum'] += item['price']
        categories_stats[cat]['count'] += 1

print("Средняя цена по категориям:")
for cat, stats in sorted(categories_stats.items()):
    if stats['count'] > 0:
        avg_price = stats['sum'] / stats['count']
        print(f"Категория {cat}: {avg_price:.2f} (на основе {stats['count']} записей)")
````

## Задание 3 (максимальная цена домена среди продающихся и не продающихся)

````python
max_price_forsale = 0.0
max_price_notforsale = 0.0

for item in all_domains:
    is_forsale = item['ForSale'].strip().lower() == 'true'
    price = item['price']

    if is_forsale:
        if price > max_price_forsale:
            max_price_forsale = price
    else:
        if price > max_price_notforsale:
            max_price_notforsale = price

print(f"Максимальная цена продающегося домена (ForSale = true): ${max_price_forsale:.2f}")
print(f"Максимальная цена непродающегося домена (ForSale = false): ${max_price_notforsale:.2f}")
````

## Задание 4 (суммарная стоимость доменов без доменного имени)

````python
total_price_no_domain = 0.0
count_no_domain = 0

for item in all_domains:
    if not item['domain'].strip():
        total_price_no_domain += item['price']
        count_no_domain += 1

print(f"Количество записей без доменного имени: {count_no_domain}")
print(f"Суммарная стоимость таких записей: ${total_price_no_domain:.2f}")
````
