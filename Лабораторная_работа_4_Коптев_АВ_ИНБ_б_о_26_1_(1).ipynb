## Решение

Файл data_v1.txt расположен репозитории. В процессе выполнения задания, выполним копирование файла в локальное хранилище Google Colab:

````python
!wget -q https://raw.githubusercontent.com/enikolaev/IT-and-Programming/main/data/data_v1.txt
````

Основной код:

````python
with open("data_v1.txt", "r", encoding="utf-8") as file:
    text = file.read()

vowels = "аеёиоуыэюяАЕЁИОУЫЭЮЯ"

max_subsequence = ""
max_length = 0

for i in range(len(text)):
    digit_sum = 0

    for j in range(i, len(text)):
        char = text[j]

        if char in vowels:
            break

        if char.isdigit():
            digit_sum += int(char)
        if digit_sum % 2 == 0:
            current = text[i:j + 1]

            if len(current) > max_length:
                max_length = len(current)
                max_subsequence = current

print("Самая длинная подпоследовательность:")
print(max_subsequence)
print("Длина:", max_length)
````
