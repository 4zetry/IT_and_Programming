## Решение

Файл **pr6.txt** расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

````python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract6/pr6.txt
````

Основной код:

````python
with open('pr6.txt', 'r') as file:
    text = file.read()

vowels = 'AEIOU'

even_digits = '02468'

count = 0

for i in range(len(text) - 1):
    if text[i] in vowels and text[i + 1] in even_digits:
        count += 1

print('Количество гласных букв, за которыми следует четное число:', count)
````
